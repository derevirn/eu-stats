```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns
from pycaret.clustering import *
mpl.rcParams['figure.dpi'] = 150

data = pd.read_csv('../data/eu_regional_data.csv')
```

```python
cluster = setup(data, normalize = True, pca = True, pca_components=6,
                remove_multicollinearity=True, multicollinearity_threshold=0.75,
                ignore_features= ['Country', 'EU Region', 'region_name'])
```

```python
model = create_model('hclust', num_clusters = 4)
```

```python
df_results = assign_model(model)
```

```python
df_dim = pd.read_csv('../data/eu_regional_data_tsne.csv', index_col = 0)
df_dim = df_dim.merge(df_results[["geo", "Cluster"]], on="geo", how="left")
```
